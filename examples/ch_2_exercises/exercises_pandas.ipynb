```python
import pandas as pd
import plotly.express as px
from hamilton import driver
```

# College DF

```python
import exercises_college_pandas
```

```python
outputs = [
    "college_raw",
    "college",
    "college_summary",
    "Top10perc",
    "scatter_matrix_top10_apps_enroll",
    "Outstate",
    "Private",
    "elite",
]

dr = driver.Builder().with_modules(exercises_college_pandas).build()
results = dr.execute(final_vars=outputs)
```

```python
results["college_raw"]
```

```python
results["college"]
```

```python
results["college_summary"]
```

```python
pd.plotting.scatter_matrix(results["scatter_matrix_top10_apps_enroll"]);
```

```python
fig = px.box(y=results["Outstate"], x=results["Private"], width=800, height=600)
fig.update_layout(
    margin={"l": 20, "r": 20, "t": 20, "b": 20},
    paper_bgcolor="LightSteelBlue",
)
```

## Feature engineer: Elite

```python
# df["Elite"] = df["Top10perc"] > 50
fig = px.box(y=results["Outstate"], x=results["elite"], width=800, height=600)
fig.update_layout(
    margin={"l": 20, "r": 20, "t": 20, "b": 20},
    paper_bgcolor="LightSteelBlue",
)
```
