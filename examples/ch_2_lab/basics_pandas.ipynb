```python
import basics_pandas
from hamilton import driver
```

# Raw data loaded

```python
conf = {"state": "raw"}
output = ["auto", "unique_horsepower"]

dr = driver.Builder().with_modules(basics_pandas).with_config(conf).build()
result = dr.execute(final_vars=output)

result = dr.execute(final_vars=output)
for k, v in result.items():
    print(k, v)
    print()

dr.visualize_execution(final_vars=output)
```

# Cleaned data loaded

```python
conf = {"state": "cleaned"}
output = ["auto", "horsepower", "unique_horsepower", "total_horsepower"]

dr = driver.Builder().with_modules(basics_pandas).with_config(conf).build()
result = dr.execute(final_vars=output)
dr.visualize_execution(final_vars=output)

result = dr.execute(final_vars=output)
for k, v in result.items():
    print(k, v)
    print()

dr.visualize_execution(final_vars=output)
```

# Visualising

```python
conf = {"state": "cleaned"}
output = [
    "auto",
]

dr = driver.Builder().with_modules(basics_pandas).with_config(conf).build()
result = dr.execute(final_vars=output)
df = result["auto"]
```

```python
df.plot.scatter("horsepower", "mpg")
```

```python
import pandas as pd
import plotly.express as px

df.cylinders = pd.Series(df.cylinders, dtype="category")
fig = px.box(df, x="cylinders", y="mpg")
fig.show()
```

```python
px.histogram(df, x="mpg", nbins=12)
```
