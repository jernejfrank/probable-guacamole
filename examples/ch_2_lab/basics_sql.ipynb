```python
tbl = db.open_table("Cars")
cars_table = tbl.to_lance()
```

```python
duckdb.query("SELECT * FROM cars_table")
```

```python
duckdb.query(
    "SELECT year"
    ", weight "
    ", origin "
    "FROM cars_table "
    "WHERE 1=1 "
    "AND year > 80 "
    "AND mpg > 30"
)
```

```python
duckdb.query(
    "SELECT name"
    ", weight "
    ", origin "
    ", displacement "
    "FROM cars_table as t "
    "WHERE 1=1 "
    "AND t.displacement < 300 "
    "AND (t.name LIKE '%ford%' OR t.name LIKE '%datsun%')"
)
```
